```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

# Display basic information
print("Dataset loaded successfully!")
print("\nShape of dataset:")
print(df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe(include="all"))
```

```python
# ============================================
# STEP 2: DATA QUALITY & EXPLORATION
# ============================================

# 1. Check missing values
print("===== MISSING VALUES =====")
missing = df.isnull().sum()

missing_df = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": (missing / len(df)) * 100
})

display(missing_df[missing_df["Missing_Count"] > 0])


# 2. Check duplicate rows
print("\n===== DUPLICATE ROWS =====")
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)


# 3. Check duplicate Farm IDs
print("\n===== DUPLICATE FARM IDs =====")
duplicate_farm_ids = df["Farm_ID"].duplicated().sum()
print("Duplicate Farm IDs:", duplicate_farm_ids)


# 4. Examine categorical columns
categorical_columns = df.select_dtypes(include="object").columns

print("\n===== CATEGORICAL COLUMNS =====")

for col in categorical_columns:
    print(f"\n{col}")
    print("Number of unique values:", df[col].nunique())
    print("Unique values:")
    print(df[col].unique())


# 5. Numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns

print("\n===== NUMERICAL COLUMNS =====")
print("Number of numerical columns:", len(numerical_columns))
print(numerical_columns.tolist())


# 6. Check for impossible / suspicious values
print("\n===== NUMERICAL VALUE RANGES =====")

range_df = pd.DataFrame({
    "Minimum": df[numerical_columns].min(),
    "Maximum": df[numerical_columns].max(),
    "Mean": df[numerical_columns].mean()
})

display(range_df)


# 7. Check number of observations per season
print("\n===== SEASON DISTRIBUTION =====")
season_counts = df["Season"].value_counts()

display(season_counts)


# 8. Check crop distribution
print("\n===== CROP DISTRIBUTION =====")
crop_counts = df["Crop"].value_counts()

display(crop_counts)


# 9. Check state distribution
print("\n===== STATE DISTRIBUTION =====")
state_counts = df["State"].value_counts()

display(state_counts)
```

```python
# ============================================
# STEP 3: DATA CLEANING
# ============================================

# Create a copy so the original dataset remains unchanged
df_clean = df.copy()

print("Original shape:", df_clean.shape)


# ------------------------------------------------
# 1. Handle missing values
# ------------------------------------------------

print("\n===== MISSING VALUES BEFORE CLEANING =====")
print(df_clean[["Rainfall_mm",
                "Soil_Moisture_pct",
                "Yield_Tonnes_Ha"]].isnull().sum())


# Use median because it is less affected by extreme values
df_clean["Rainfall_mm"] = df_clean["Rainfall_mm"].fillna(
    df_clean["Rainfall_mm"].median()
)

df_clean["Soil_Moisture_pct"] = df_clean["Soil_Moisture_pct"].fillna(
    df_clean["Soil_Moisture_pct"].median()
)

df_clean["Yield_Tonnes_Ha"] = df_clean["Yield_Tonnes_Ha"].fillna(
    df_clean["Yield_Tonnes_Ha"].median()
)


# ------------------------------------------------
# 2. Verify missing values
# ------------------------------------------------

print("\n===== MISSING VALUES AFTER CLEANING =====")

missing_after = df_clean.isnull().sum()

display(
    missing_after[missing_after > 0]
)

if missing_after.sum() == 0:
    print("No missing values remaining.")
else:
    print("Some missing values still exist.")


# ------------------------------------------------
# 3. Check duplicates again
# ------------------------------------------------

print("\n===== DUPLICATE CHECK =====")

print("Duplicate rows:", df_clean.duplicated().sum())
print("Duplicate Farm IDs:", df_clean["Farm_ID"].duplicated().sum())


# ------------------------------------------------
# 4. Check impossible values
# ------------------------------------------------

print("\n===== VALIDITY CHECKS =====")

print("Farm area <= 0:",
      (df_clean["Farm_Area_Hectares"] <= 0).sum())

print("Rainfall < 0:",
      (df_clean["Rainfall_mm"] < 0).sum())

print("Humidity outside 0-100:",
      ((df_clean["Humidity_pct"] < 0) |
       (df_clean["Humidity_pct"] > 100)).sum())

print("Soil moisture outside 0-100:",
      ((df_clean["Soil_Moisture_pct"] < 0) |
       (df_clean["Soil_Moisture_pct"] > 100)).sum())

print("Seed quality outside 0-1:",
      ((df_clean["Seed_Quality_Score"] < 0) |
       (df_clean["Seed_Quality_Score"] > 1)).sum())

print("Disease/pest risk outside 0-100:",
      ((df_clean["Disease_Pest_Risk_pct"] < 0) |
       (df_clean["Disease_Pest_Risk_pct"] > 100)).sum())

print("Profit calculations checked.")


# ------------------------------------------------
# 5. Validate financial relationship
# ------------------------------------------------

df_clean["Calculated_Profit"] = (
    df_clean["Revenue_INR"] - df_clean["Total_Cost_INR"]
)

profit_difference = (
    df_clean["Profit_INR"] - df_clean["Calculated_Profit"]
)

print("\n===== FINANCIAL VALIDATION =====")

print("Maximum difference between recorded and calculated profit:",
      profit_difference.abs().max())

print("Rows with inconsistent profit:",
      (profit_difference.abs() > 1).sum())


# ------------------------------------------------
# 6. Validate production relationship
# ------------------------------------------------

df_clean["Calculated_Production"] = (
    df_clean["Farm_Area_Hectares"] *
    df_clean["Yield_Tonnes_Ha"]
)

production_difference = (
    df_clean["Production_Tonnes"] -
    df_clean["Calculated_Production"]
)

print("\n===== PRODUCTION VALIDATION =====")

print("Maximum production difference:",
      production_difference.abs().max())

print("Rows with production difference > 0.1 tonnes:",
      (production_difference.abs() > 0.1).sum())


# ------------------------------------------------
# 7. Final shape
# ------------------------------------------------

print("\n===== FINAL DATASET =====")
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])

print("\nCleaning step completed.")
```

```python
# ============================================
# STEP 4: SEASONAL EXPLORATORY DATA ANALYSIS
# ============================================

# Remove temporary validation columns
df_clean = df_clean.drop(
    columns=["Calculated_Profit", "Calculated_Production"]
)

print("Dataset shape after removing temporary columns:")
print(df_clean.shape)


# ------------------------------------------------
# 1. Number of farms in each season
# ------------------------------------------------

print("\n===== FARMS BY SEASON =====")

season_count = (
    df_clean["Season"]
    .value_counts()
    .rename_axis("Season")
    .reset_index(name="Farm_Count")
)

display(season_count)


# ------------------------------------------------
# 2. Average agricultural performance by season
# ------------------------------------------------

print("\n===== AGRICULTURAL PERFORMANCE BY SEASON =====")

season_performance = (
    df_clean
    .groupby("Season")
    .agg(
        Farms=("Farm_ID", "count"),
        Avg_Yield=("Yield_Tonnes_Ha", "mean"),
        Avg_Production=("Production_Tonnes", "mean"),
        Avg_Revenue=("Revenue_INR", "mean"),
        Avg_Cost=("Total_Cost_INR", "mean"),
        Avg_Profit=("Profit_INR", "mean"),
        Avg_Water_Used=("Water_Used_m3", "mean"),
        Avg_Water_Efficiency=(
            "Water_Efficiency_t_per_1000m3",
            "mean"
        ),
        Avg_Disease_Pest_Risk=(
            "Disease_Pest_Risk_pct",
            "mean"
        )
    )
    .round(2)
)

display(season_performance)


# ------------------------------------------------
# 3. Environmental conditions by season
# ------------------------------------------------

print("\n===== ENVIRONMENTAL CONDITIONS BY SEASON =====")

season_environment = (
    df_clean
    .groupby("Season")
    .agg(
        Avg_Rainfall=("Rainfall_mm", "mean"),
        Avg_Temperature=("Avg_Temperature_C", "mean"),
        Avg_Humidity=("Humidity_pct", "mean"),
        Avg_Sunlight=("Sunlight_Hours_Day", "mean"),
        Avg_Soil_Moisture=("Soil_Moisture_pct", "mean"),
        Avg_Soil_pH=("Soil_pH", "mean")
    )
    .round(2)
)

display(season_environment)


# ------------------------------------------------
# 4. Resource usage by season
# ------------------------------------------------

print("\n===== RESOURCE USAGE BY SEASON =====")

season_resources = (
    df_clean
    .groupby("Season")
    .agg(
        Avg_Nitrogen=("Nitrogen_kg_ha", "mean"),
        Avg_Phosphorus=("Phosphorus_kg_ha", "mean"),
        Avg_Potassium=("Potassium_kg_ha", "mean"),
        Avg_Fertilizer=("Fertilizer_kg_ha", "mean"),
        Avg_Pesticide=("Pesticide_Litre_ha", "mean"),
        Avg_Water_Used=("Water_Used_m3", "mean")
    )
    .round(2)
)

display(season_resources)


# ------------------------------------------------
# 5. Profitability distribution
# ------------------------------------------------

print("\n===== PROFITABILITY BY SEASON =====")

profitability = (
    df_clean
    .groupby("Season")
    .agg(
        Average_Profit=("Profit_INR", "mean"),
        Median_Profit=("Profit_INR", "median"),
        Minimum_Profit=("Profit_INR", "min"),
        Maximum_Profit=("Profit_INR", "max"),
        Profitable_Farms=(
            "Profit_INR",
            lambda x: (x > 0).sum()
        ),
        Loss_Making_Farms=(
            "Profit_INR",
            lambda x: (x < 0).sum()
        )
    )
    .round(2)
)

display(profitability)


# ------------------------------------------------
# 6. Profitability percentage
# ------------------------------------------------

profitability["Profitability_Rate_%"] = (
    profitability["Profitable_Farms"]
    / (
        profitability["Profitable_Farms"]
        + profitability["Loss_Making_Farms"]
    )
    * 100
).round(2)

print("\n===== PROFITABILITY RATE =====")
display(profitability[["Profitable_Farms",
                       "Loss_Making_Farms",
                       "Profitability_Rate_%"]])


print("\nStep 4 completed.")
```

```python
# ============================================
# STEP 5: VISUALIZATION OF SEASONAL PATTERNS
# ============================================

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")


# ------------------------------------------------
# 1. Number of farms by season
# ------------------------------------------------

plt.figure(figsize=(8, 5))

sns.barplot(
    data=season_count,
    x="Season",
    y="Farm_Count"
)

plt.title("Number of Farms by Season")
plt.xlabel("Season")
plt.ylabel("Number of Farms")

plt.show()


# ------------------------------------------------
# 2. Average Yield by Season
# ------------------------------------------------

yield_plot = (
    season_performance
    .reset_index()
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=yield_plot,
    x="Season",
    y="Avg_Yield"
)

plt.title("Average Agricultural Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")

plt.show()


# ------------------------------------------------
# 3. Average Profit by Season
# ------------------------------------------------

plt.figure(figsize=(8, 5))

sns.barplot(
    data=yield_plot,
    x="Season",
    y="Avg_Profit"
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")

plt.show()


# ------------------------------------------------
# 4. Profitability Rate
# ------------------------------------------------

profit_rate_plot = (
    profitability
    .reset_index()
)

plt.figure(figsize=(8, 5))

sns.barplot(
    data=profit_rate_plot,
    x="Season",
    y="Profitability_Rate_%"
)

plt.title("Percentage of Profitable Farms by Season")
plt.xlabel("Season")
plt.ylabel("Profitable Farms (%)")

plt.ylim(0, 100)

plt.show()


# ------------------------------------------------
# 5. Water Usage vs Water Efficiency
# ------------------------------------------------

water_plot = (
    season_performance
    .reset_index()
)

fig, ax1 = plt.subplots(figsize=(8, 5))

sns.barplot(
    data=water_plot,
    x="Season",
    y="Avg_Water_Used",
    ax=ax1
)

ax1.set_title("Water Usage and Water Efficiency by Season")
ax1.set_xlabel("Season")
ax1.set_ylabel("Average Water Used (m³)")


ax2 = ax1.twinx()

sns.lineplot(
    data=water_plot,
    x="Season",
    y="Avg_Water_Efficiency",
    marker="o",
    linewidth=2,
    ax=ax2
)

ax2.set_ylabel(
    "Water Efficiency (Tonnes per 1000 m³)"
)

plt.show()


# ------------------------------------------------
# 6. Environmental Conditions
# ------------------------------------------------

environment_plot = (
    season_environment
    .reset_index()
)

environment_melted = environment_plot.melt(
    id_vars="Season",
    var_name="Environmental_Factor",
    value_name="Average_Value"
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=environment_melted,
    x="Environmental_Factor",
    y="Average_Value",
    hue="Season"
)

plt.title("Environmental Conditions Across Seasons")
plt.xlabel("Environmental Factor")
plt.ylabel("Average Value")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()


print("Step 5 completed.")
```

```python
# ============================================
# STEP 6: CROP × SEASON ANALYSIS
# ============================================


# ------------------------------------------------
# 1. Number of farms for each Crop × Season
# ------------------------------------------------

print("===== FARM COUNT: CROP × SEASON =====")

crop_season_count = (
    df_clean
    .groupby(["Crop", "Season"])
    .size()
    .unstack(fill_value=0)
)

display(crop_season_count)


# ------------------------------------------------
# 2. Average Yield: Crop × Season
# ------------------------------------------------

print("\n===== AVERAGE YIELD: CROP × SEASON =====")

crop_season_yield = (
    df_clean
    .groupby(["Crop", "Season"])["Yield_Tonnes_Ha"]
    .mean()
    .unstack()
    .round(2)
)

display(crop_season_yield)


# ------------------------------------------------
# 3. Average Profit: Crop × Season
# ------------------------------------------------

print("\n===== AVERAGE PROFIT: CROP × SEASON =====")

crop_season_profit = (
    df_clean
    .groupby(["Crop", "Season"])["Profit_INR"]
    .mean()
    .unstack()
    .round(2)
)

display(crop_season_profit)


# ------------------------------------------------
# 4. Average Revenue: Crop × Season
# ------------------------------------------------

print("\n===== AVERAGE REVENUE: CROP × SEASON =====")

crop_season_revenue = (
    df_clean
    .groupby(["Crop", "Season"])["Revenue_INR"]
    .mean()
    .unstack()
    .round(2)
)

display(crop_season_revenue)


# ------------------------------------------------
# 5. Average Water Efficiency: Crop × Season
# ------------------------------------------------

print("\n===== WATER EFFICIENCY: CROP × SEASON =====")

crop_season_water = (
    df_clean
    .groupby(["Crop", "Season"])
    ["Water_Efficiency_t_per_1000m3"]
    .mean()
    .unstack()
    .round(2)
)

display(crop_season_water)


# ------------------------------------------------
# 6. Complete Crop × Season Performance Table
# ------------------------------------------------

print("\n===== COMPLETE CROP × SEASON PERFORMANCE =====")

crop_season_summary = (
    df_clean
    .groupby(["Crop", "Season"])
    .agg(
        Farms=("Farm_ID", "count"),
        Avg_Yield=("Yield_Tonnes_Ha", "mean"),
        Avg_Production=("Production_Tonnes", "mean"),
        Avg_Revenue=("Revenue_INR", "mean"),
        Avg_Cost=("Total_Cost_INR", "mean"),
        Avg_Profit=("Profit_INR", "mean"),
        Avg_Water_Used=("Water_Used_m3", "mean"),
        Avg_Water_Efficiency=(
            "Water_Efficiency_t_per_1000m3",
            "mean"
        ),
        Avg_Disease_Risk=(
            "Disease_Pest_Risk_pct",
            "mean"
        )
    )
    .round(2)
)

display(crop_season_summary)


# ------------------------------------------------
# 7. Best season for each crop based on yield
# ------------------------------------------------

print("\n===== BEST SEASON FOR EACH CROP (BY YIELD) =====")

best_season_by_yield = (
    crop_season_yield
    .idxmax(axis=1)
    .to_frame(name="Best_Season")
)

best_season_by_yield["Highest_Avg_Yield"] = (
    crop_season_yield.max(axis=1)
)

display(best_season_by_yield)


# ------------------------------------------------
# 8. Best season for each crop based on profit
# ------------------------------------------------

print("\n===== BEST SEASON FOR EACH CROP (BY PROFIT) =====")

best_season_by_profit = (
    crop_season_profit
    .idxmax(axis=1)
    .to_frame(name="Most_Profitable_Season")
)

best_season_by_profit["Highest_Avg_Profit"] = (
    crop_season_profit.max(axis=1)
)

display(best_season_by_profit)


# ------------------------------------------------
# 9. Best crop in each season based on yield
# ------------------------------------------------

print("\n===== BEST CROP IN EACH SEASON (BY YIELD) =====")

best_crop_by_season = (
    crop_season_yield
    .idxmax(axis=0)
    .to_frame(name="Best_Crop")
)

best_crop_by_season["Highest_Avg_Yield"] = (
    crop_season_yield.max(axis=0)
)

display(best_crop_by_season)


# ------------------------------------------------
# 10. Best crop in each season based on profit
# ------------------------------------------------

print("\n===== BEST CROP IN EACH SEASON (BY PROFIT) =====")

best_crop_profit = (
    crop_season_profit
    .idxmax(axis=0)
    .to_frame(name="Best_Crop")
)

best_crop_profit["Highest_Avg_Profit"] = (
    crop_season_profit.max(axis=0)
)

display(best_crop_profit)


print("\nStep 6 completed.")
```

```python
# ============================================
# STEP 7: CROP × SEASON HEATMAP ANALYSIS
# ============================================

import matplotlib.pyplot as plt
import seaborn as sns


# ------------------------------------------------
# 1. Yield Heatmap
# ------------------------------------------------

plt.figure(figsize=(10, 6))

sns.heatmap(
    crop_season_yield,
    annot=True,
    fmt=".2f",
    linewidths=0.5
)

plt.title("Average Yield by Crop and Season")
plt.xlabel("Season")
plt.ylabel("Crop")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 2. Profit Heatmap
# ------------------------------------------------

plt.figure(figsize=(10, 6))

sns.heatmap(
    crop_season_profit,
    annot=True,
    fmt=",.0f",
    linewidths=0.5,
    center=0
)

plt.title("Average Profit by Crop and Season")
plt.xlabel("Season")
plt.ylabel("Crop")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 3. Water Efficiency Heatmap
# ------------------------------------------------

plt.figure(figsize=(10, 6))

sns.heatmap(
    crop_season_water,
    annot=True,
    fmt=".2f",
    linewidths=0.5
)

plt.title("Water Efficiency by Crop and Season")
plt.xlabel("Season")
plt.ylabel("Crop")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 4. Revenue Heatmap
# ------------------------------------------------

plt.figure(figsize=(10, 6))

sns.heatmap(
    crop_season_revenue,
    annot=True,
    fmt=",.0f",
    linewidths=0.5
)

plt.title("Average Revenue by Crop and Season")
plt.xlabel("Season")
plt.ylabel("Crop")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 5. Find highest and lowest crop-season
#    combinations
# ------------------------------------------------

print("===== EXTREME CROP × SEASON COMBINATIONS =====")


# Yield
yield_stack = crop_season_yield.stack()

print("\nHighest Yield:")
print(yield_stack.idxmax(), "=", round(yield_stack.max(), 2))

print("\nLowest Yield:")
print(yield_stack.idxmin(), "=", round(yield_stack.min(), 2))


# Profit
profit_stack = crop_season_profit.stack()

print("\nHighest Profit:")
print(profit_stack.idxmax(), "=", round(profit_stack.max(), 2))

print("\nLowest Profit:")
print(profit_stack.idxmin(), "=", round(profit_stack.min(), 2))


# Water efficiency
water_stack = crop_season_water.stack()

print("\nHighest Water Efficiency:")
print(water_stack.idxmax(), "=", round(water_stack.max(), 2))

print("\nLowest Water Efficiency:")
print(water_stack.idxmin(), "=", round(water_stack.min(), 2))


# Revenue
revenue_stack = crop_season_revenue.stack()

print("\nHighest Revenue:")
print(revenue_stack.idxmax(), "=", round(revenue_stack.max(), 2))

print("\nLowest Revenue:")
print(revenue_stack.idxmin(), "=", round(revenue_stack.min(), 2))


print("\nStep 7 completed.")
```

```python
# ============================================
# STEP 8: STATE × SEASON ANALYSIS
# ============================================


# ------------------------------------------------
# 1. Farm count by State × Season
# ------------------------------------------------

print("===== FARM COUNT: STATE × SEASON =====")

state_season_count = (
    df_clean
    .groupby(["State", "Season"])
    .size()
    .unstack(fill_value=0)
)

display(state_season_count)


# ------------------------------------------------
# 2. Average Yield by State × Season
# ------------------------------------------------

print("\n===== AVERAGE YIELD: STATE × SEASON =====")

state_season_yield = (
    df_clean
    .groupby(["State", "Season"])["Yield_Tonnes_Ha"]
    .mean()
    .unstack()
    .round(2)
)

display(state_season_yield)


# ------------------------------------------------
# 3. Average Profit by State × Season
# ------------------------------------------------

print("\n===== AVERAGE PROFIT: STATE × SEASON =====")

state_season_profit = (
    df_clean
    .groupby(["State", "Season"])["Profit_INR"]
    .mean()
    .unstack()
    .round(2)
)

display(state_season_profit)


# ------------------------------------------------
# 4. Average Revenue by State × Season
# ------------------------------------------------

print("\n===== AVERAGE REVENUE: STATE × SEASON =====")

state_season_revenue = (
    df_clean
    .groupby(["State", "Season"])["Revenue_INR"]
    .mean()
    .unstack()
    .round(2)
)

display(state_season_revenue)


# ------------------------------------------------
# 5. Water Efficiency by State × Season
# ------------------------------------------------

print("\n===== WATER EFFICIENCY: STATE × SEASON =====")

state_season_water = (
    df_clean
    .groupby(
        ["State", "Season"]
    )["Water_Efficiency_t_per_1000m3"]
    .mean()
    .unstack()
    .round(2)
)

display(state_season_water)


# ------------------------------------------------
# 6. Disease/Pest Risk by State × Season
# ------------------------------------------------

print("\n===== DISEASE/PEST RISK: STATE × SEASON =====")

state_season_risk = (
    df_clean
    .groupby(["State", "Season"])
    ["Disease_Pest_Risk_pct"]
    .mean()
    .unstack()
    .round(2)
)

display(state_season_risk)


# ------------------------------------------------
# 7. Best season for each state based on yield
# ------------------------------------------------

print("\n===== BEST SEASON FOR EACH STATE (BY YIELD) =====")

best_state_season_yield = (
    state_season_yield
    .idxmax(axis=1)
    .to_frame(name="Best_Season")
)

best_state_season_yield["Highest_Avg_Yield"] = (
    state_season_yield.max(axis=1)
)

display(best_state_season_yield)


# ------------------------------------------------
# 8. Best season for each state based on profit
# ------------------------------------------------

print("\n===== BEST SEASON FOR EACH STATE (BY PROFIT) =====")

best_state_season_profit = (
    state_season_profit
    .idxmax(axis=1)
    .to_frame(name="Best_Season")
)

best_state_season_profit["Highest_Avg_Profit"] = (
    state_season_profit.max(axis=1)
)

display(best_state_season_profit)


# ------------------------------------------------
# 9. Best state in each season based on yield
# ------------------------------------------------

print("\n===== BEST STATE IN EACH SEASON (BY YIELD) =====")

best_state_yield = (
    state_season_yield
    .idxmax(axis=0)
    .to_frame(name="Best_State")
)

best_state_yield["Highest_Avg_Yield"] = (
    state_season_yield.max(axis=0)
)

display(best_state_yield)


# ------------------------------------------------
# 10. Best state in each season based on profit
# ------------------------------------------------

print("\n===== BEST STATE IN EACH SEASON (BY PROFIT) =====")

best_state_profit = (
    state_season_profit
    .idxmax(axis=0)
    .to_frame(name="Best_State")
)

best_state_profit["Highest_Avg_Profit"] = (
    state_season_profit.max(axis=0)
)

display(best_state_profit)


print("\nStep 8 completed.")
```

```python
# ============================================
# STEP 10: STATISTICAL SIGNIFICANCE TESTING
# ============================================

import pandas as pd
import numpy as np

from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd


# ------------------------------------------------
# 1. Prepare groups
# ------------------------------------------------

seasons = ["Kharif", "Rabi", "Zaid"]

yield_groups = [
    df_clean.loc[
        df_clean["Season"] == season,
        "Yield_Tonnes_Ha"
    ].dropna()
    for season in seasons
]

profit_groups = [
    df_clean.loc[
        df_clean["Season"] == season,
        "Profit_INR"
    ].dropna()
    for season in seasons
]

water_groups = [
    df_clean.loc[
        df_clean["Season"] == season,
        "Water_Efficiency_t_per_1000m3"
    ].dropna()
    for season in seasons
]


# ------------------------------------------------
# 2. Descriptive statistics
# ------------------------------------------------

print("===== DESCRIPTIVE STATISTICS BY SEASON =====")

season_stats = (
    df_clean
    .groupby("Season")
    .agg(
        Yield_Mean=("Yield_Tonnes_Ha", "mean"),
        Yield_Median=("Yield_Tonnes_Ha", "median"),
        Yield_Std=("Yield_Tonnes_Ha", "std"),

        Profit_Mean=("Profit_INR", "mean"),
        Profit_Median=("Profit_INR", "median"),
        Profit_Std=("Profit_INR", "std"),

        Water_Efficiency_Mean=(
            "Water_Efficiency_t_per_1000m3",
            "mean"
        ),
        Water_Efficiency_Median=(
            "Water_Efficiency_t_per_1000m3",
            "median"
        ),
        Water_Efficiency_Std=(
            "Water_Efficiency_t_per_1000m3",
            "std"
        )
    )
    .reindex(seasons)
    .round(3)
)

display(season_stats)


# ------------------------------------------------
# 3. One-way ANOVA: Yield
# ------------------------------------------------

print("\n===== ONE-WAY ANOVA: YIELD =====")

yield_anova = stats.f_oneway(
    *yield_groups
)

print("F-statistic:", round(yield_anova.statistic, 4))
print("p-value:", yield_anova.pvalue)


# ------------------------------------------------
# 4. One-way ANOVA: Profit
# ------------------------------------------------

print("\n===== ONE-WAY ANOVA: PROFIT =====")

profit_anova = stats.f_oneway(
    *profit_groups
)

print("F-statistic:", round(profit_anova.statistic, 4))
print("p-value:", profit_anova.pvalue)


# ------------------------------------------------
# 5. One-way ANOVA: Water Efficiency
# ------------------------------------------------

print("\n===== ONE-WAY ANOVA: WATER EFFICIENCY =====")

water_anova = stats.f_oneway(
    *water_groups
)

print("F-statistic:", round(water_anova.statistic, 4))
print("p-value:", water_anova.pvalue)


# ------------------------------------------------
# 6. Kruskal-Wallis: Yield
# ------------------------------------------------

print("\n===== KRUSKAL-WALLIS TEST: YIELD =====")

yield_kw = stats.kruskal(
    *yield_groups
)

print("H-statistic:", round(yield_kw.statistic, 4))
print("p-value:", yield_kw.pvalue)


# ------------------------------------------------
# 7. Kruskal-Wallis: Profit
# ------------------------------------------------

print("\n===== KRUSKAL-WALLIS TEST: PROFIT =====")

profit_kw = stats.kruskal(
    *profit_groups
)

print("H-statistic:", round(profit_kw.statistic, 4))
print("p-value:", profit_kw.pvalue)


# ------------------------------------------------
# 8. Kruskal-Wallis: Water Efficiency
# ------------------------------------------------

print("\n===== KRUSKAL-WALLIS TEST: WATER EFFICIENCY =====")

water_kw = stats.kruskal(
    *water_groups
)

print("H-statistic:", round(water_kw.statistic, 4))
print("p-value:", water_kw.pvalue)


# ------------------------------------------------
# 9. Effect size: Eta Squared
# ------------------------------------------------

def eta_squared(groups):
    """
    Calculate eta squared for one-way ANOVA.
    """

    all_values = np.concatenate(groups)

    grand_mean = np.mean(all_values)

    ss_between = sum(
        len(group) * (np.mean(group) - grand_mean) ** 2
        for group in groups
    )

    ss_total = sum(
        (all_values - grand_mean) ** 2
    )

    return ss_between / ss_total


yield_eta = eta_squared(yield_groups)
profit_eta = eta_squared(profit_groups)
water_eta = eta_squared(water_groups)


print("\n===== EFFECT SIZE: ETA SQUARED =====")

print("Yield η²:", round(yield_eta, 4))
print("Profit η²:", round(profit_eta, 4))
print("Water Efficiency η²:", round(water_eta, 4))


# ------------------------------------------------
# 10. Tukey HSD: Yield
# ------------------------------------------------

print("\n===== TUKEY HSD: YIELD =====")

yield_tukey = pairwise_tukeyhsd(
    endog=df_clean["Yield_Tonnes_Ha"],
    groups=df_clean["Season"],
    alpha=0.05
)

print(yield_tukey)


# ------------------------------------------------
# 11. Tukey HSD: Profit
# ------------------------------------------------

print("\n===== TUKEY HSD: PROFIT =====")

profit_tukey = pairwise_tukeyhsd(
    endog=df_clean["Profit_INR"],
    groups=df_clean["Season"],
    alpha=0.05
)

print(profit_tukey)


# ------------------------------------------------
# 12. Tukey HSD: Water Efficiency
# ------------------------------------------------

print("\n===== TUKEY HSD: WATER EFFICIENCY =====")

water_tukey = pairwise_tukeyhsd(
    endog=df_clean["Water_Efficiency_t_per_1000m3"],
    groups=df_clean["Season"],
    alpha=0.05
)

print(water_tukey)


print("\nStep 10 completed.")
```

```python
# ============================================
# STEP 11: TWO-WAY ANOVA
# STATE × SEASON INTERACTION
# ============================================

import pandas as pd
import numpy as np
import statsmodels.api as sm

from statsmodels.formula.api import ols


# ------------------------------------------------
# Function to perform two-way ANOVA
# ------------------------------------------------

def run_two_way_anova(data, outcome):

    formula = (
        f"{outcome} ~ C(State) + C(Season) "
        f"+ C(State):C(Season)"
    )

    model = ols(formula, data=data).fit()

    anova_table = sm.stats.anova_lm(
        model,
        typ=2
    )

    return model, anova_table


# ------------------------------------------------
# 1. Yield
# ------------------------------------------------

print("============================================")
print("TWO-WAY ANOVA: YIELD")
print("============================================")

yield_model, yield_anova = run_two_way_anova(
    df_clean,
    "Yield_Tonnes_Ha"
)

display(yield_anova.round(6))


# ------------------------------------------------
# 2. Profit
# ------------------------------------------------

print("\n============================================")
print("TWO-WAY ANOVA: PROFIT")
print("============================================")

profit_model, profit_anova = run_two_way_anova(
    df_clean,
    "Profit_INR"
)

display(profit_anova.round(6))


# ------------------------------------------------
# 3. Water Efficiency
# ------------------------------------------------

print("\n============================================")
print("TWO-WAY ANOVA: WATER EFFICIENCY")
print("============================================")

water_model, water_anova = run_two_way_anova(
    df_clean,
    "Water_Efficiency_t_per_1000m3"
)

display(water_anova.round(6))


# ------------------------------------------------
# 4. Extract p-values
# ------------------------------------------------

print("\n============================================")
print("P-VALUES")
print("============================================")

results = pd.DataFrame({
    "Outcome": [
        "Yield",
        "Profit",
        "Water Efficiency"
    ],

    "State_p_value": [
        yield_anova.loc["C(State)", "PR(>F)"],
        profit_anova.loc["C(State)", "PR(>F)"],
        water_anova.loc["C(State)", "PR(>F)"]
    ],

    "Season_p_value": [
        yield_anova.loc["C(Season)", "PR(>F)"],
        profit_anova.loc["C(Season)", "PR(>F)"],
        water_anova.loc["C(Season)", "PR(>F)"]
    ],

    "Interaction_p_value": [
        yield_anova.loc[
            "C(State):C(Season)",
            "PR(>F)"
        ],

        profit_anova.loc[
            "C(State):C(Season)",
            "PR(>F)"
        ],

        water_anova.loc[
            "C(State):C(Season)",
            "PR(>F)"
        ]
    ]
})

display(results)


# ------------------------------------------------
# 5. Partial Eta Squared
# ------------------------------------------------

def partial_eta_squared(anova_table, effect):

    ss_effect = anova_table.loc[
        effect,
        "sum_sq"
    ]

    ss_error = anova_table.loc[
        "Residual",
        "sum_sq"
    ]

    return ss_effect / (ss_effect + ss_error)


print("\n============================================")
print("PARTIAL ETA SQUARED")
print("============================================")


effects = [
    "C(State)",
    "C(Season)",
    "C(State):C(Season)"
]

partial_eta_results = []

for effect in effects:

    partial_eta_results.append({
        "Effect": effect,

        "Yield": partial_eta_squared(
            yield_anova,
            effect
        ),

        "Profit": partial_eta_squared(
            profit_anova,
            effect
        ),

        "Water_Efficiency": partial_eta_squared(
            water_anova,
            effect
        )
    })


partial_eta_df = pd.DataFrame(
    partial_eta_results
).set_index("Effect")

display(partial_eta_df.round(4))


# ------------------------------------------------
# 6. Identify whether interaction is significant
# ------------------------------------------------

print("\n============================================")
print("INTERACTION SIGNIFICANCE")
print("============================================")

interaction_results = []

for outcome, table in [
    ("Yield", yield_anova),
    ("Profit", profit_anova),
    ("Water Efficiency", water_anova)
]:

    p_value = table.loc[
        "C(State):C(Season)",
        "PR(>F)"
    ]

    interaction_results.append({
        "Outcome": outcome,
        "Interaction_p_value": p_value,
        "Significant_at_0.05": p_value < 0.05
    })


interaction_df = pd.DataFrame(
    interaction_results
)

display(interaction_df)


print("\nStep 11 completed.")
```
